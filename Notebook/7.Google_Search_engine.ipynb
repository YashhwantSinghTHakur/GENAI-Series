```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_groq import ChatGroq
from langchain.agents import create_agent
from langgraph.checkpoint.memory import MemorySaver
```

```python
llm = ChatGroq(model="openai/gpt-oss-120b", streaming = True
           )
search  = GoogleSerperAPIWrapper()
```

```python
agent = create_agent(
    model = llm,
    tools = [search.run],
    system_prompt="you are agent and can search for my question on google",
    checkpointer=MemorySaver()
)
```

```python
question = "what is the address of TechSimPlus learnings ?"
```

```python
response = agent.invoke({"messages":[{"role":"user","content":question}]},
                        {"configurable":{"thread_id": "xyc"}})
```

```python
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_number(a:int, b:int):
    """
       It will return the sum of two number
       aegs:
          a : number a
          b : number b

    """    
    return a + b
```

```python
@tool
def multiply_number(a:int, b:int):
    """
       It will return the multiply of two number
       aegs:
          a : number a
          b : number b

    """    
    return a * b
```

```python
add_number.invoke({"a" : 22, "b": 34})
```

```python
from langchain_groq import ChatGroq


from langchain.agents import create_agent
```

```python
llm = ChatGroq(model="qwen/qwen3.8-27b", 
           )
```

```python
agent = create_agent(
    model=llm,
    tools=[add_number, multiply_number],
    system_prompt= "you are a math teacher, and always use tool for calculation"
)
```

```python
res = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "what is 2 + 3 * 4 ?"
        }
    ]
})

print(res)
```
